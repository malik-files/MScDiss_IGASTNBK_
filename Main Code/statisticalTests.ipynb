```python
import numpy as np
from statsmodels.stats.proportion import proportions_ztest
```

```python
#We are comparing the men to women
#We are comparing wikipedia values to grammy values
maleCount = np.array([107024,137])
femaleCount = np.array([40582,40])
totalMenWomen = np.array([147606,177])
test_stat, p_val = proportions_ztest(count = maleCount, nobs = totalMenWomen, alternative = 'two-sided' )
print('Two sided z-test: z = {:.4f}, p value = {:.4f}'.format(test_stat, p_val))
test_stat, p_val = proportions_ztest(count = femaleCount, nobs = totalMenWomen, alternative = 'two-sided' )
print('Two sided z-test for females: z = {:.4f}, p value = {:.4f}'.format(test_stat, p_val))
#I have to accept the null hypothesis
```
