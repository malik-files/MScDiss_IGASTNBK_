```python
import pandas as pd
import numpy as np
import networkx as nx
from func.UsefulFunctions import *
```

```python
groupDF = pd.read_parquet("./Data/Network_groupWikidata.parquet")

groupDF.head()
```

```python
graphInfluences = addToNetwork4(groupDF, "influencedby", "group", "has influenced", "influence", "Musical Group", "influencedbyLabel", "groupLabel")
```

```python
graphRecordLabel = addToNetwork4(groupDF, "recordLabel", "group", "has/had group", "Record Label", "Musical Group","recordLabelLabel", "groupLabel")
```

```python
graphGenre = addToNetwork4(groupDF, "genre", "group", "has been participated in by", "Genre", "Musical Artist", "genreLabel", "groupLabel")
```

```python
firstGraph = nx.compose(graphInfluences, graphGenre)
groupGraph = nx.compose(firstGraph, graphRecordLabel)
```

```python
genreGraph = nx.read_graphml("./graphs/groupsGraph.graphml")
groupGraph = nx.compose(genreGraph, groupGraph)
```

```python
#Get only british artists
britishGroups = groupDF[(groupDF["countryOfOriginLabel"] == "Northern Ireland") | (groupDF["countryOfOriginLabel"] == "Scotland")| (groupDF["countryOfOriginLabel"] == "England") | (groupDF["countryOfOriginLabel"] == "Wales") | (groupDF["countryOfOriginLabel"] == "United Kingdom") | (groupDF["countryOfOriginLabel"] == "United Kingdom of Great Britain and Northern Ireland") | (groupDF["countryOfOriginLabel"] == "United Kingdom of Great Britain and Ireland")]

display(britishGroups)
```

```python
graphInfluencesUK = addToNetwork4(britishGroups, "influencedby", "group", "has influenced", "influence", "Musical Group",
                                "influencedbyLabel", "groupLabel")
graphRecordLabelUK = addToNetwork4(britishGroups, "recordLabel", "group", "has/had group", "Record Label", "Musical Group",
                                 "recordLabelLabel", "groupLabel")
graphGenreUK = addToNetwork4(britishGroups, "genre", "group", "has been participated in by", "Genre", "Musical Artist",
                           "genreLabel", "groupLabel")
firstGraphUK = nx.compose(graphInfluencesUK, graphGenreUK)
ukGroupGraph = nx.compose(firstGraphUK, graphRecordLabelUK)
ukGroupGraph = nx.compose(genreGraph, ukGroupGraph)
```
