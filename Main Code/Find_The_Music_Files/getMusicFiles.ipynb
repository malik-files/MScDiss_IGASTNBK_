```python
def isFileFormatInString(item):
    listOfAudioFileFormats = ['.ogg','.oga','.opus''.mp3','.aac','.wma','.flac','.alac','.wav','.mid','.aiff','.dsf','.dff']
    for x in listOfAudioFileFormats:
        if x in item.lower():
            return True
    return False

def isFileFormatInStringNEW(item):
    #Get the last element of the list which should be the file extension
    fileFormat = (item.split('.')[-1]).lower()
    #print(fileFormat)
    listOfAudioFileFormats = ['ogg','oga','opus','mp3','aac','wma','flac','alac','wav','mid','aiff','dsf','dff']
    if fileFormat in listOfAudioFileFormats:
        return True
    return False



```

```python
#!/usr/bin/python3
import requests
import pandas as pd
import numpy as np

def getAllFiles(title):

    """
    This code has been adapted from the example code from Mediawiki.
    Accessed on 2/07/2026 from https://www.mediawiki.org/wiki/API:Images

    Original Text below: 
    get_page_images.py

    MediaWiki API Demos
    Demo of `Images` module: Send a GET request to obtain a JSON
	object listing all of the image files embedded on a single page

    MIT License

"""


    S = requests.Session()
    URL = "https://en.wikipedia.org/w/api.php"
    headers = {'User-Agent': 'MAlikDiss/1.0 (aja25@st-andrews.ac.uk)'}

    PARAMS = {
        "action": "query",
        "format": "json",
        "imlimit": 'max',
        "titles": title,
        "prop": "images"
    }

    try:
        R = S.get(url=URL, params=PARAMS, headers=headers)
        DATA = R.json()

        PAGES = DATA['query']['pages']
        files = set()

        for k, v in PAGES.items():
            if "images" in v.keys():
                for img in v['images']:
                    if isFileFormatInStringNEW(img["title"]):
                        files.add(img["title"])
            else:
                continue
    except:
        return "none"

    if len(files) > 0:
        return files
    else:
        return "none"
```

```python
getAllFiles("Drum and bass")
```

```python
def writeToFile(listOfTitles):
    noAudioGenres = set()
    genresWithAudios = set()
    with open("List of articles with the sound files.txt","+w") as f:
        for x in listOfTitles:
            result = getAllFiles(x)
            f.write(x.upper() + "\n"+ str(result) +"\n")

            if result == "none":
                noAudioGenres.add(x)
            else:
                genresWithAudios.add(x)
    
    return genresWithAudios, noAudioGenres



```

```python
genreCSV = "/Users/malikawotorebo/Documents/GitHub/MSc Dissertation/Find_The_Music_Files/Query 19 - Genres on wikidata and their wikipedia pages.csv"
genreTable = pd.read_csv(genreCSV)
genreTable.head()
genreArticleTitles = genreTable["page_titleEN"]
genresWithAudios, genresWithoutAudio = writeToFile(genreArticleTitles)
print(f"{len(genresWithAudios)} genre articles have audio files within them, and {len(genresWithoutAudio)} do not")
```
