```python
def fileFormatsUsed(item):
    if item is not None:
        fileFormat = item.split('.')[-1]
        return fileFormat
```

```python
#!/usr/bin/python3
import requests
import pandas as pd
import numpy as np

def getAllFiles(title):

    """
    This code has been adapted from the example code from Mediawiki.
    Accessed on 2/07/2026 from https://www.mediawiki.org/wiki/API:Images

    Original Text below: 
    get_page_images.py

    MediaWiki API Demos
    Demo of `Images` module: Send a GET request to obtain a JSON
	object listing all of the image files embedded on a single page

    MIT License

"""


    S = requests.Session()
    URL = "https://en.wikipedia.org/w/api.php"
    headers = {'User-Agent': 'MAlikDiss/1.0 (aja25@st-andrews.ac.uk)'}

    PARAMS = {
        "action": "query",
        "format": "json",
        "imlimit": 'max',
        "titles": title,
        "prop": "images"
    }

    try:
        R = S.get(url=URL, params=PARAMS, headers=headers)
        DATA = R.json()

        PAGES = DATA['query']['pages']
        files = set()

        for k, v in PAGES.items():
            if "images" in v.keys():
                for img in v['images']:
                    files.add(img["title"])
            else:
                continue
    except:
        return None

    if len(files) > 0:
        return files
    else:
        return None
```

```python
def getAllFileFormats(listOfTitles):
    fileExtensions = set()
    for x in listOfTitles:
        files = getAllFiles(x)
        if files is not None:
            for file in files:
                fileExtensions.add(fileFormatsUsed(file))
    
    return fileExtensions
```

```python
genreCSV = "/Users/malikawotorebo/Documents/GitHub/MSc Dissertation/Find_The_Music_Files/Query 19 - Genres on wikidata and their wikipedia pages.csv"
genreTable = pd.read_csv(genreCSV)
genreTable.head()
genreArticleTitles = genreTable["page_titleEN"]
fileExt = getAllFileFormats(genreArticleTitles)
fileExtList = list(fileExt)
fileExtList.sort()
with open("list of file extensions.txt","+w") as f:
    for x in fileExtList:
        f.write(x + "\n")
```
